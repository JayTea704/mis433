## Question 4. Adding new columns

Currently, dep_time and sched_dep_time are convenient to look at, but hard to compute with because they're not continuous numbers.

For example, 759 means 7:59 and 801 means 8:01. Their difference is not 42 but 2 minutes.

```python
# Convert them to a more convenient representation: the number of minutes since midnight (0).
flights['dep_time_mins'] = (flights['dep_time'] // 100 * 60 + flights['dep_time'] % 100) % 1440
flights['sched_dep_time_mins'] = (flights['sched_dep_time'] // 100 * 60 + flights['sched_dep_time'] % 100) % 1440

flights
```

```python
# Create a new column for arr_time - dep_time.
flights['arr_time_mins'] = (flights['arr_time'] // 100 * 60 + flights['arr_time'] % 100) % 1440

flights['travel_time'] = flights['arr_time_mins'] - flights['dep_time_mins']



# Compare this column with air_time.
flights['time_diff'] = flights['travel_time'] - flights['air_time']

flights['time_diff'].describe()
```

```python
# Compare dep_time, sched_dep_time, and dep_delay. How would you expect those three numbers to be related?
# Try creating a column to calculate dep_delay from dep_time and sched_dep_time (and/or other columns if necessary).
flights['dep_delay_calc'] = flights['dep_time_mins'] - flights['sched_dep_time_mins']

flights.loc[flights['dep_delay_calc'] < -60,'dep_delay_calc'] += 1440

# Test your results.
wrong = flights[(flights['dep_delay_calc'] != flights['dep_delay']) & flights['dep_delay'].notna()]

len(wrong)
```

## Question 5. Mixing things together

The following questions may require multiple operations above.

```python
# Find the 20 most delayed flights.
# Display the following: year, month, day, carrier, flight, dep_delay, and arr_delay.
# How do you want to handle ties?
# according to stackoverflow usng method = 'min' makes the tied flights all the lowest rank in the group


cols = ['year', 'month', 'day', 'carrier', 'flight', 'dep_delay', 'arr_delay']

flights['delay_rank'] = flights['dep_delay'].rank(method='min', ascending=False)

flights[flights['delay_rank'] <= 20].sort_values('delay_rank')[cols + ['delay_rank']]
```

```python
# Sort all AA flights to find the top 10 fastest (highest speed) flights.
# Display the following: year, month, day, carrier, flight, origin, dest, distance, air_time, and speed (miles per hour).
aa = flights[flights['carrier'] == 'AA'].copy()

aa['speed'] = aa['distance'] / (aa['air_time'] / 60)

aa['speed_rank'] = aa['speed'].rank(method='min', ascending=False)

cols = ['year', 'month', 'day', 'carrier', 'flight', 'origin', 'dest',
        'distance', 'air_time', 'speed']

aa[aa['speed_rank'] <= 10].sort_values('speed_rank')[cols + ['speed_rank']]
```

```python
# Find all flights that satisfy the following:
# - From John F. Kennedy Airport (JFK) or Newark Airport (EWR) to Seattle-Tacoma Airport (SEA).
# - The carrier is UA, AA, or DL.
# - Dates from 4/1/2013 (inclusive) through 4/3/2013 (inclusive).
# - Scheduled arrival time is before noon.
# - Display the following: year, month, day, carrier, flight, origin, dest, sched_dep_time, and sched_arr_time.
# - Sort by year, month, day, sched_arr_time

cols = ['year', 'month', 'day', 'carrier', 'flight', 'origin', 'dest',
        'sched_dep_time', 'sched_arr_time']

result = flights[
    flights['origin'].isin(['JFK', 'EWR']) &
    (flights['dest'] == 'SEA') &
    flights['carrier'].isin(['UA', 'AA', 'DL']) &
    (flights['year'] == 2013) &
    (flights['month'] == 4) &
    flights['day'].between(1, 3) &
    (flights['sched_arr_time'] < 1200)
]

result[cols].sort_values(['year', 'month', 'day', 'sched_arr_time'])
```
